```python
import numpy as np
import tensorflow  as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,SimpleRNN,Dense

```

```python
word_index =imdb.get_word_index()
reverse_word_index = {value:key  for key,value in word_index.items()}
```

```python
from tensorflow.keras.models import load_model

model = load_model('simple_rnn_imdb.h5')
model.summary()
```

```python
model.get_weights()
```

```python
import re

def decode_review(encoded_review):
    return ' '.join(reverse_word_index.get(index - 3, '?') for index in encoded_review)

def preprocess_text(text):
    cleaned_text = re.sub(r'[^a-zA-Z\s]', '', text.lower())
    words = cleaned_text.split()
    encoded_review = [word_index.get(word, 2) + 3 for word in words]
    return sequence.pad_sequences([encoded_review], maxlen=500)
```

```python
## Prediction function
def prediction_sentiment(review):
    preprocessed_input = preprocess_text(review)
    prediction = model.predict(preprocessed_input, verbose=0)
    score = float(prediction[0][0])
    sentiment = 'Positive' if score >= 0.49 else 'Negative'
    return sentiment, score
```

```python
example_review = "This movie was fantastic! The acting was great and The plot was thrilling."

sentiment, score = prediction_sentiment(example_review)

print(f"Review: {example_review}")
print(f"Sentiment: {sentiment}")
print(f"Prediction score: {score:.4f}")
```
