```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sent=[ 'the glass of milk',
      'the glass of juice',
      'the cup of tea',
      'i am a good developar',
      'understand the meaning of words',
      'your video are good'    
]
```

```python
sent
```

```python
# define vocabulary size 
voc_size = 10000
```

```python
## one_hot representation of the sentences
one_hot_repr=[one_hot(words,voc_size)for words in sent]

one_hot_repr    
    
```

```python
## word Emebedding representation
from tensorflow.keras.layers import Embedding
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
import numpy as np
```

```python
sent_lenght = 8
embedding_dco = pad_sequences(one_hot_repr,padding='pre',maxlen=sent_lenght)
embedding_dco
```

```python
## feature representation
dim=10
```

```python
model=Sequential()
model.add(Embedding(voc_size, dim, input_length=sent_lenght))
model.compile('adam', 'mse')
```

```python
model.summary()
```
